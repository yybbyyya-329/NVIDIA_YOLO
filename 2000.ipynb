```python
from ultralytics import YOLO
import cv2
import os
import numpy as np

# 1. 모델 로드
model = YOLO('runs/pose/train10/weights/best.pt')
class_names = model.names  # ['person']

# 2. 이미지 폴더 경로
image_folder = 'dataset/train/images'
output_folder = 'keypoint_output'

# 3. 출력 폴더 생성
os.makedirs(output_folder, exist_ok=True)

# 4. COCO skeleton 연결 정의
skeleton = [
    (0, 1), (1, 2), (2, 3), (3, 4),
    (1, 5), (5, 6), (6, 7),
    (1, 8), (8, 9), (9, 10),
    (8, 11), (11, 12)
]

# 5. 이미지 순회하며 추론
for img_name in os.listdir(image_folder):
    img_path = os.path.join(image_folder, img_name)

    if img is None:
        print(f"🚫 {img_name}는 열 수 없습니다.")
        continue

    results = model(img)
    output = img.copy()

    if results and results[0].keypoints is not None and results[0].keypoints.xy is not None:
        boxes = results[0].boxes.xyxy.cpu().numpy()
        kpts = results[0].keypoints.xy.cpu().numpy()
        confs = results[0].boxes.conf.cpu().numpy()
        cls_ids = results[0].boxes.cls.cpu().numpy().astype(int)

        total = min(len(boxes), len(kpts), len(confs), len(cls_ids))

        for i in range(total):
            box = boxes[i]
            person = kpts[i]
            conf = confs[i]
            cls_id = cls_ids[i]

            x1, y1, x2, y2 = box
            label = f"{class_names[cls_id]} {conf:.2f}" if cls_id < len(class_names) else f"cls {cls_id} {conf:.2f}"

            # 바운딩 박스
            cv2.rectangle(output, (int(x1), int(y1)), (int(x2), int(y2)), (255, 255, 0), 2)

            # 라벨
            (tw, th), _ = cv2.getTextSize(label, cv2.FONT_HERSHEY_SIMPLEX, 0.6, 1)
            cv2.rectangle(output, (int(x1), int(y1) - th - 6), (int(x1) + tw, int(y1)), (255, 255, 0), -1)
            cv2.putText(output, label, (int(x1), int(y1) - 4), cv2.FONT_HERSHEY_SIMPLEX, 0.6, (0, 0, 0), 1)

            # 관절 연결
            for pt1, pt2 in skeleton:
                try:
                    px1, py1 = person[pt1]
                    px2, py2 = person[pt2]
                    if all([px1 > 0, py1 > 0, px2 > 0, py2 > 0]):
                        cv2.line(output, (int(px1), int(py1)), (int(px2), int(py2)), (0, 255, 0), 2)
                except IndexError:
                    continue

            # 관절 점
            for x, y in person:
                cv2.circle(output, (int(x), int(y)), 3, (0, 0, 255), -1)

    # 6. 결과 저장
    save_path = os.path.join(output_folder, img_name)
    cv2.imwrite(save_path, output)
    print(f"✅ 저장 완료: {save_path}")

```

```python
from ultralytics import YOLO
import cv2
import numpy as np
import traceback

# 1. 포즈 모델 로드
model = YOLO('C:/Users/user/Desktop/a_trffic_signal.v1i.yolov8/runs/pose/train10/weights/best.pt')

# 2. 클래스 이름 (YOLO 모델에 저장된 클래스명 사용)
class_names = model.names  # ['person'] 등

# 3. 비디오 로딩
cap = cv2.VideoCapture('42.mp4')
width = int(cap.get(cv2.CAP_PROP_FRAME_WIDTH))
height = int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))
fourcc = cv2.VideoWriter_fourcc(*'mp4v')
out = cv2.VideoWriter('key_point_vodo2.mp4', fourcc, 30.0, (width, height))

# 4. 관절 연결 쌍 (COCO 포맷)
skeleton = [
    (0, 1), (1, 2), (2, 3), (3, 4),
    (1, 5), (5, 6), (6, 7),
    (1, 8), (8, 9), (9, 10),
    (8, 11), (11, 12)
]

while cap.isOpened():
    ret, frame = cap.read()
    if not ret:
        break

    results = model.predict(frame)
    output = frame.copy()

    if results and results[0].keypoints is not None and results[0].keypoints.xy is not None:
        try:
            boxes = results[0].boxes.xyxy.cpu().numpy()
            kpts = results[0].keypoints.xy.cpu().numpy()
            confs = results[0].boxes.conf.cpu().numpy()
            cls_ids = results[0].boxes.cls.cpu().numpy().astype(int)

            total = min(len(boxes), len(kpts), len(confs), len(cls_ids))

            for i in range(total):
                box = boxes[i]
                person = kpts[i]
                conf = confs[i]
                cls_id = cls_ids[i]

                x1, y1, x2, y2 = box
                if cls_id < len(class_names):
                    label = f"{class_names[cls_id]} {conf:.2f}"
                else:
                    label = f"cls {cls_id} {conf:.2f}"

                # 박스
                cv2.rectangle(output, (int(x1), int(y1)), (int(x2), int(y2)), (255, 255, 0), 2)

                # 라벨
                (tw, th), _ = cv2.getTextSize(label, cv2.FONT_HERSHEY_SIMPLEX, 0.6, 1)
                cv2.rectangle(output, (int(x1), int(y1) - th - 6), (int(x1) + tw, int(y1)), (255, 255, 0), -1)
                cv2.putText(output, label, (int(x1), int(y1) - 4), cv2.FONT_HERSHEY_SIMPLEX, 0.6, (0, 0, 0), 1)

                # 관절 연결
                for pt1, pt2 in skeleton:
                    try:
                        px1, py1 = person[pt1]
                        px2, py2 = person[pt2]
                        if all([px1 > 0, py1 > 0, px2 > 0, py2 > 0]):
                            cv2.line(output, (int(px1), int(py1)), (int(px2), int(py2)), (0, 255, 0), 2)
                    except IndexError:
                        continue

                # 점
                for x, y in person:
                    cv2.circle(output, (int(x), int(y)), 3, (0, 0, 255), -1)
        except Exception as e:
            print(f"⚠️ 처리 중 예외 발생: {e}")
            traceback.print_exc()

    out.write(output)
    cv2.imshow('Pose with Box + Label', output)

    if cv2.waitKey(1) & 0xFF == ord('q'):
        break

cap.release()
out.release()
cv2.destroyAllWindows()
```
