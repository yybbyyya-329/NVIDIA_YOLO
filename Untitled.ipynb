```python
import cv2
import os
from glob import glob

img_folder = 'runs\pose\predict'
output_video = 'pose_result2.mp4'

# 이미지 목록
img_files = sorted(glob(os.path.join(img_folder, '*.jpg')) + glob(os.path.join(img_folder, '*.png')))
if not img_files:
    print("❌ 결과 이미지가 없습니다.")
    exit()

# 비디오 설정
frame = cv2.imread(img_files[0])
height, width, _ = frame.shape
video = cv2.VideoWriter(output_video, cv2.VideoWriter_fourcc(*'mp4v'), 5, (width, height))

for img in img_files:
    video.write(cv2.imread(img))

video.release()
print(f" 비디오 저장 완료: {output_video}")
```

```python
from ultralytics import YOLO
import cv2

# 학습된 포즈 모델 로드 (task='pose' 제거)
model_path = 'C:/Users/user/Desktop/a_trffic_signal.v1i.yolov8/runs/pose/train7/weights/best.pt'
model = YOLO(model_path)

# 비디오 경로
video_path = 'C:/Users/choes/Desktop/a_trffic_signal.v1i.yolov8/33.mp4'
cap = cv2.VideoCapture(video_path)

# 출력 비디오 설정
fourcc = cv2.VideoWriter_fourcc(*'mp4v')
out = cv2.VideoWriter('output_video.mp4', fourcc, 30.0, (int(cap.get(3)), int(cap.get(4))))

while cap.isOpened():
    ret, frame = cap.read()
    if not ret:
        break

    # 키포인트 추론
    results = model(frame)

    # 키포인트가 시각화된 프레임 얻기
    annotated_frame = results[0].plot()

    # 화면에 표시 + 저장
    cv2.imshow('Pose Estimation', annotated_frame)
    out.write(annotated_frame)

    if cv2.waitKey(1) & 0xFF == ord('q'):
        break

cap.release()
out.release()
cv2.destroyAllWindows()
```

```python
from ultralytics import YOLO
import cv2

# 모델 로드 (task는 자동 인식)
model = YOLO('C:/Users/user/Desktop/a_trffic_signal.v1i.yolov8/runs/pose/train7/weights/best.pt')

# 비디오 로드
video_path = 'C:/Users/choes/Desktop/a_trffic_signal.v1i.yolov8/33.mp4'
cap = cv2.VideoCapture(video_path)

# 출력 설정
fourcc = cv2.VideoWriter_fourcc(*'mp4v')
out = cv2.VideoWriter('output_video.mp4', fourcc, 30.0, (int(cap.get(3)), int(cap.get(4))))

while cap.isOpened():
    ret, frame = cap.read()
    if not ret:
        break

    # 👇 YOLO는 BGR → RGB로 바꿔줘야 정확
    rgb_frame = cv2.cvtColor(frame, cv2.COLOR_BGR2RGB)

    # 👇 추론 (results는 리스트 형태)
    results = model.predict(source=rgb_frame, save=False, verbose=False)

    # 👇 추론 결과 시각화
    annotated_frame = results[0].plot()

    # 👇 BGR로 다시 변환 후 저장/표시
    annotated_frame = cv2.cvtColor(annotated_frame, cv2.COLOR_RGB2BGR)
    out.write(annotated_frame)
    cv2.imshow('Keypoint Pose', annotated_frame)

    if cv2.waitKey(1) & 0xFF == ord('q'):
        break

cap.release()
out.release()
cv2.destroyAllWindows()
```

```python
from ultralytics import YOLO
import cv2

# 1. 공식 포즈 모델 로딩
model = YOLO('C:/Users/user/Desktop/a_trffic_signal.v1i.yolov8/runs/pose/train7/weights/best.pt')  # 네 모델도 사용 가능

# 2. 비디오 로딩
cap = cv2.VideoCapture('33.mp4')
fourcc = cv2.VideoWriter_fourcc(*'mp4v')
out = cv2.VideoWriter('pose_output.mp4', fourcc, 30.0, (int(cap.get(3)), int(cap.get(4))))

while cap.isOpened():
    ret, frame = cap.read()
    if not ret:
        break

    # 3. 추론
    results = model.predict(frame)

    # 4. 시각화 및 저장
    annotated = results[0].plot()
    out.write(annotated)
    cv2.imshow('Pose', annotated)

    if cv2.waitKey(1) & 0xFF == ord('q'):
        break

cap.release()
out.release()
cv2.destroyAllWindows()
```
