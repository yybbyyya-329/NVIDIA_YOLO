```python
# 1. YOLOv8 설치
!pip install -q ultralytics

# 2. 라이브러리 임포트
from ultralytics import YOLO
from google.colab import files
import cv2
import numpy as np
import matplotlib.pyplot as plt
from PIL import Image

# 3. YOLOv8x 모델 로드
model = YOLO('yolov8x.pt')  # 사전 학습된 COCO 모델

# 4. 이미지 업로드 (Colab에서 실행 시 파일 선택창 뜸)
uploaded = files.upload()

# 5. 업로드한 이미지 파일 경로 얻기
image_path = list(uploaded.keys())[0]

# 6. 이미지 열기
image = Image.open(image_path)

# 7. 객체 감지 수행
results = model(image)

# 8. 'cat' 클래스만 카운트
cat_count = 0
for result in results:
    for box in result.boxes:
        cls_id = int(box.cls[0])
        cls_name = model.names[cls_id]
        if cls_name.lower() == 'cat':
            cat_count += 1

# 9. 결과 출력
print(f"🔍 감지된 고양이 수: {cat_count}")

# 10. 감지된 객체 시각화 (matplotlib 사용)
results[0].show()

# 추가: 결과 이미지 저장하고 보기
results[0].save(filename="detected.jpg")
display(Image.open("detected.jpg"))
```
