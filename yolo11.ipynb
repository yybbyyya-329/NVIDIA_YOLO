```python
import cv2
from ultralytics import YOLO
from google.colab import files
from IPython.display import display, Video
import os

# YOLO11 모델 로드
model = YOLO("yolo11n.pt")

# 영상 파일 업로드
uploaded = files.upload()
video_path = list(uploaded.keys())[0]  # 업로드된 파일 이름 가져오기

# 영상 파일 열기
cap = cv2.VideoCapture(video_path)
if not cap.isOpened():
    print("Error: Could not open video file.")
    raise SystemExit

# 출력 영상 설정
output_path = "output_video.mp4"
fourcc = cv2.VideoWriter_fourcc(*"mp4v")
fps = int(cap.get(cv2.CAP_PROP_FPS))
width = int(cap.get(cv2.CAP_PROP_FRAME_WIDTH))
height = int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))
out = cv2.VideoWriter(output_path, fourcc, fps, (width, height))

# 자동차 카운트 변수
car_count = 0
tracked_ids = set()

# 영상 처리
while cap.isOpened():
    ret, frame = cap.read()
    if not ret:
        break

    # YOLO11로 객체 감지 및 추적
    results = model.track(frame, persist=True)

    for result in results:
        boxes = result.boxes
        for box in boxes:
            class_id = int(box.cls[0])
            conf = float(box.conf[0])
            if class_id == 2 and conf > 0.5:  # 자동차(class_id=2)만 감지
                x1, y1, x2, y2 = map(int, box.xyxy[0])
                track_id = int(box.id[0]) if box.id is not None else None

                if track_id and track_id not in tracked_ids:
                    tracked_ids.add(track_id)
                    car_count += 1

                # 바운딩 박스와 라벨 그리기
                cv2.rectangle(frame, (x1, y1), (x2, y2), (0, 255, 0), 2)
                label = f"Car {track_id} ({conf:.2f})"
                cv2.putText(frame, label, (x1, y1 - 10), cv2.FONT_HERSHEY_SIMPLEX, 0.5, (0, 255, 0), 2)

    # 총 자동차 수 표시
    cv2.putText(frame, f"Total Cars: {car_count}", (10, 30), cv2.FONT_HERSHEY_SIMPLEX, 1, (0, 0, 255), 2)

    # 프레임 저장
    out.write(frame)

# 리소스 해제
cap.release()
out.release()

print(f"Total number of cars detected: {car_count}")

# 완성된 영상 Colab에서 표시
if os.path.exists(output_path):
    display(Video(output_path, embed=True))
else:
    print("Error: Output video was not created.")

# 다운로드 옵션 제공
files.download(output_path)
```
